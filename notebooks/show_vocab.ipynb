# Browse the vocabulary

(Notebook version of `scripts/show_vocab.py`.)

The tokenizer's vocabulary is a table of 49,152 `token string → id` entries. This notebook prints a slice of
it so you can see what the tokens look like: special tokens first, then punctuation, digits and single letters, then common words and
pieces. `Ġ` means a leading space. Special tokens such as `<|endoftext|>` (id 0) are included.

Select the kernel **smalllm (.venv)** before running.

```python
from huggingface_hub import hf_hub_download
from tokenizers import Tokenizer

tok = Tokenizer.from_file(hf_hub_download("HuggingFaceTB/SmolLM2-135M", "tokenizer.json"))
```

## Choose a slice
`COUNT` tokens starting at id `START`. Change these and re-run the next cell.

```python
COUNT = 50
START = 0

# with_added_tokens=True includes special tokens such as <|endoftext|> (id 0).
vocab = tok.get_vocab(with_added_tokens=True)  # dict: token string -> id
by_id = sorted(vocab.items(), key=lambda item: item[1])
for token, token_id in by_id[START : START + COUNT]:
    print(f"{token_id:>6}  {token!r}")
```

## Look further in
Common whole words live in the thousands; the last ids are rare pieces.

```python
for token, token_id in by_id[6400:6410] + by_id[-5:]:
    print(f"{token_id:>6}  {token!r}")
```

## Look up a single token either way

```python
print("id of 'Ġthe':", tok.token_to_id("Ġthe"))
print("token 0:     ", tok.id_to_token(0))
```
